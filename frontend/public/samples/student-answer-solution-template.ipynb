## Question 1
Explain the difference between supervised and unsupervised learning.

Supervised learning trains on labeled data where each input has a known output. The model learns a mapping from inputs to labels and is commonly used for classification and regression.

Unsupervised learning trains on unlabeled data and tries to discover structure or patterns, such as clusters or lower-dimensional representations. Typical examples include clustering and dimensionality reduction.

## Question 2
Write a Python function to compute factorial of n.

```python
def factorial(n):
    if n < 0:
        raise ValueError('n must be non-negative')
    if n in (0, 1):
        return 1

    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial(5))  # 120
```
